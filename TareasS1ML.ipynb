# Tarea 1
## Función que muestra la tabla de multiplicar.

```python
def tabla_multiplicarV1(n, x):
    for i in range(1, x + 1):
        print(f"{n} x {i} = {n * i}")

n = int(input(f"Ingrese la tabla que quiere: "))
x = int(input(f"¿Hasta qué valor?: "))

tabla_multiplicarV1(n, x)
```

# Tarea 2
## Sistema de inicio de sesión

```python
# Usuario y contraseña correctos
usuario_correcto = "usuario22"
password_correcto = "01020304"

intentos = 3

for i in range(3):

    usuario = input("Ingrese su usuario: ")
    password = input("Ingrese su contraseña: ")

    if usuario == usuario_correcto and password == password_correcto:
        print("Ingreso correcto")
        break

    else:
        intentos -= 1

        if intentos > 0:
            print("Usuario o contraseña incorrectos")
            print("Intentos restantes:", intentos)
        else:
            print("Cuenta bloqueada")
```
